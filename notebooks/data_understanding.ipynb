```python
# sku_master column names
sku_master.columns
```

```python
# Checking null values
sku_master.isnull().sum()
```

```python
# Checking duplicate values
sku_master.duplicated().sum()
```

```python
# Checking Category Count
category_count = sku_master.groupby("Category")["SKU"].count()
print(category_count)
```

```python
# Checking Subcategory Count
subcategory_count = sku_master.groupby("Subcategory")["SKU"].count()
print(subcategory_count)
```

```python
# Finding SKUs Where Selling Price is Lower Than cost Price
below_cost = sku_master[
    sku_master["Selling_Price"]< sku_master["Cost_Price"]
]
print(below_cost)
```

```python
# Converting Launch_Date str to datetime
sku_master['Launch_Date'] = pd.to_datetime(sku_master["Launch_Date"])
sku_master.dtypes
```

```python
# Checking All SKUs Whose launch date > last date in our sales data
sku_master[sku_master["Launch_Date"]> pd.Timestamp("2025-12-31")]
```

```python
# checking SKUs has a future launch date relative to the beginning of the sales period
sku_master[sku_master["Launch_Date"]< pd.Timestamp("2024-01-01")]
```

```python
# Check whether Gross Margin is calculated correctly
expected_margin = sku_master["Selling_Price"] - sku_master["Cost_Price"]

# Find rows where the stored margin doesn't match the calculated margin
margin_check = sku_master[
    abs(expected_margin - sku_master["Gross_Margin_Per_Unit"]) > 0.01
]

print(margin_check)
print("Number of inconsistent rows:", len(margin_check))
```

## SKU Master Data Quality Findings

- SKU master contains 50 unique SKUs with no duplicate SKU IDs.
- There are no missing values in the SKU master data.
- The 50 SKUs are evenly distributed across 5 categories, with 10 SKUs in each category.
- There are 10 subcategories, with 5 SKUs in each.
- The `Launch_Date` column was converted to datetime format.
- 16 SKUs have a selling price lower than their cost price, resulting in negative gross margin.
- No SKU has a launch date after the end of the sales data period.
- The calculated gross margin matches the stored `Gross_Margin_Per_Unit` values.

```python
# Convert date column from str to datetime
calendar['date'] = pd.to_datetime(calendar['date'])
calendar['date'].dtype
```

```python
# Checking Duplicates
calendar['date'].duplicated().sum()
```

```python
# Getting minimum and maximum date
print("Minimum date:", calendar["date"].min())
print("Maximum date:", calendar["date"].max())
```

```python
# Check for gaps between consecutive calendar dates
date_diff = calendar["date"].diff()

# Find dates where the gap is not exactly 1 day
missing_date_gaps = calendar[date_diff != pd.Timedelta(days=1)]

print("Number of date gaps:", len(missing_date_gaps))
print(missing_date_gaps)
```

```python
# Checking Unique Days of Week
calendar['day_of_week'].unique()
```

```python
# Identify rows where the weekend flag does not match the day of week
expected_weekend = calendar["day_of_week"].isin(["Saturday", "Sunday"])

inconsistent_weekend = calendar[
    expected_weekend !=(calendar["is_weekend"] == 1)
]

print("Number of inconsistent weekend rows:", len(inconsistent_weekend))
print(inconsistent_weekend)
```

```python
# Check whether the holiday flag matches the holiday name
expected_holiday = calendar["holiday"].notnull()

inconsistent_holiday = calendar[
    expected_holiday != calendar["is_holiday"]
]

print("Number of inconsistent holiday rows:", len(inconsistent_holiday))
print(inconsistent_holiday)
```

```python
# Check valid month values
calendar["month"].unique()
```

```python
# Check valid quarter values
calendar["quarter"].unique()
```

```python
# Check the range of week values
print("Minimum week:", calendar["week"].min())
print("Maximum week:", calendar["week"].max())
```

```python
# Check valid weekend flag values
calendar["is_weekend"].unique()
```

```python
# Check valid holiday flag values
calendar["is_holiday"].unique()
```

## Calendar Data Quality Findings

- Calendar data contains 731 rows covering 2024-01-01 to 2025-12-31.
- There are no duplicate dates and no missing dates in the sequence.
- The `date` column was converted to datetime format.
- `is_weekend` and `is_holiday` values match the actual day and holiday information.
- Month, quarter, week, weekend, and holiday values are valid.
